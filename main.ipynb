```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import re as re

df = pd.read_csv("smartphones_uncleaned.csv")
df
```

```python
df.isnull().sum()[lambda x: x>0]
```

```python
df.drop(columns=['review_link', 'expert_rating', 'review'], inplace = True)
```

```python

df['user_rating'] = df['user_rating'].fillna("Unknown")
df['processor'] = df['processor'].fillna("Unknown")
df['ram_internal_memory'] = df['ram_internal_memory'].fillna(df['ram_internal_memory'].mode()[0])
df['operating_system'] = df['operating_system'].fillna(df['operating_system'].mode()[0])
df['additional_features'] = df['additional_features'].fillna("None")
df['front_cameras'] = df['front_cameras'].fillna(df['front_cameras'].mode()[0])
df.isnull().sum()[lambda x:x > 0]
```

```python
df
```

```python
df['price'] = df['price'].str.replace(',','')
df['price'] = df['price'].str.strip()

df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['price'] = df['price'].astype(int)

df['price']
```

```python
df.info()
```

```python
split_data = df['ram_internal_memory'].str.split('|', expand=True)

df['ram_gb'] = split_data[0].str.extract(r'(\d+)').astype(float)
df['storage'] = split_data[1].str.extract(r'(\d+)').astype(float)
```

```python
df.drop(columns=['ram_internal_memory'])
```

```python
df['price'] = df['price'] * 0.011
df['price']
```

```python
df = df.rename(columns= {
    'price' : 'Price_USD',
    'ram_gb' : 'Ram_Capacity',
    'Storage' : 'Internal Storage'
})
```

```python
new_OS = df['operating_system'].str.split(' ', expand=True)
df['OS'] = new_OS[0].str.extract(r'(Android|IOS|BlackBerry|Sailfish|Tizen|Windows)', expand=False, flags=re.IGNORECASE)

df['OS'].value_counts()
```

```python
df.drop(columns=['operating_system'])
```

```python
df.info()
```

```python
df.drop(columns=['ram_internal_memory'])
```

```python
battery_capacity = df['battery'].str.split('|', expand=True)
df['Battery_Capacity'] = battery_capacity[0].str.extract(r'(\d+)').astype(float)
```

```python
df['Battery_Capacity'] = df['Battery_Capacity'].astype(int)
```

```python
df.info()
```

```python
df.drop(columns=['ram_internal_memory', 'operating_system', 'battery'])
```

```python
df['display'].unique()
```

```python
df['Display'] = df['display'].str.extract(r'(AMOLED|OLED|LCD|IPS|TFT)')
df['Display']

df['Refresh_Rate'] = df['display'].str.extract(r'(\d+)Hz').astype(float)
df['Screen_Size'] = df['display'].str.extract(r'(\d+\.?\d*)''').astype(float)
```

```python
df.info()
```

```python
df
```

```python
df.isnull().sum()[lambda x:x>0]
```

```python
df['storage']= df['storage'].fillna(df['storage'].mode()[0])
```

```python
df.isnull().sum()[lambda x:x>0]
```

```python
df['OS'] = df['OS'].fillna("Other")
df['Refresh_Rate'] = df['Refresh_Rate'].fillna('Unknown')
df['Display'] = df['Display'].fillna('Other')
df['Screen_Size'] = df['Screen_Size'].fillna('Unknown')
```

```python
df.info()
```

```python
pd.set_option('display.max_rows', None) # Tampilkan semua baris
print(df['processor'].value_counts())
pd.reset_option('display.max_rows')

df['Chipset'] = df['processor'].str.extract(r'(Snapdragon|MediaTek|Dimensity|Helio|Exynos|Apple|A\d+|Bionic|Google|Tensor|Unisoc|Kirin|HiSilicon|Spreadtrum|Marvell|core|Core|Quad|Dual|Single)')

mapping = {
    'Dimensity' :'MediaTek',
    'Helio' : 'MediaTek',
    'Kirin' : 'HiSilicon',
    'Bionic' : 'Apple A-Series',
    'Apple' : 'Apple A-Series',
    'Google' : 'Google Tensor',
    'Tensor' : 'Google Tensor',
    'Quad' : 'Core Series',
    'Dual' : 'Core Series',
    'Single':'Core Series',
    'Core' : 'Core Series',
    'core' : 'Core Series'
}
df['Chipset'] = df['Chipset'].replace(mapping)
df['Chipset'] = df['Chipset'].fillna('Unknown')
```

```python
df = df.drop(columns=['ram_internal_memory', 'battery', 'processor', 'operating_system', 'display'])
```

```python
df['Chipset'] = df['Chipset'].replace('Spreadtrum', 'Unisoc')
df['Chipset'] = df['Chipset'].replace('Unknown', 'Others')
```
