```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("cleaned_smartphones.csv")
df['Chipset'] = df['Chipset'].replace('HiSilicon', 'Unisoc')
df
```

```python
plt.figure(figsize=(12,6))

ax = sns.countplot(df, x='Chipset', palette='magma',width=1)
for container in ax.containers:
    ax.bar_label(container, padding=3)
# plt.xticks(rotation=40)
plt.title("Jumlah Handphone berdasarkan Chipset")
```

```python
plt.figure(figsize=(12,6))

ax = sns.countplot(df, x='Display', palette='magma',width=1)
for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.title("Jumlah Handphone berdasarkan Display")
```

```python

exynos = df[(df['Display'] != 'AMOLED') & (df['Chipset'] == 'Exynos')]['model_name'].unique().tolist()
exynos
```

```python
snapdragon_amoled = df[(df['Chipset'] == 'Snapdragon') & (df['Display'] == 'AMOLED')]['model_name'].unique().tolist()
snapdragon_amoled
```

```python
df['Brand'] = df['model_name'].str.extract(r'(vivo|Samsung|Motorola|Moto|Asus|OnePlus|CMF|Oppo|OPPO|Realme|realme|Infinix|IQOO|Iphone|Xiaomi|LG|Honor|Google|Pixel|Nokia|Nubia|Sony|Apple|Nothing|POCO|Tecno|Itel)')

mapping = {
    'Moto' : 'Motorola',
    'vivo' : 'Vivo',
    'Google' : 'Google Pixel',
    'Pixel' : 'Google Pixel',
    'realme' :'Realme',
    'Oppo' :'OPPO',
    'Apple' :'Iphone'
}

df['Brand'] = df['Brand'].replace(mapping)
df['Brand'] = df['Brand'].fillna('Others')
```

```python
df['Brand'].value_counts()
```

```python
sns.set_style('darkgrid')
plt.figure(figsize=(12,6))
plt.title("Rata Rata harga handphone berdasarkan chipset")
sns.lineplot(df, x='Chipset', y='Price_USD', palette='mako')
```

```python
plt.figure(figsize=(17,7))
plt.title("Rata Rata harga smartphone per brand")
ax = sns.barplot(df, x='Brand', y='Price_USD', width=0.4, errorbar=None, palette='mako')
for container in ax.containers :
    ax.bar_label(container, padding=5, fmt="%.1f")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
